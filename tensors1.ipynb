## Comparing in tensors

##### greater than
##### less than
##### equal to
##### equal to or less
##### equal to or great
##### not equal to

```python
a = tt.tensor([1,2,3])
b = a.clone() + 5  #copy the a and also add 5 to each element of the b
print(a)
print(b)
```

```python
print(a>b) #false in all elements  greater than
```

```python
a<b #true true true (in all elements) - less than
```

```python
a==b #a is equal to b which is false
```

```python
a!=b # a isn't equal to b that is true
```

```python
a>=b # a is greater than b or equal to b which is false
```

```python
a<=b # a is less than b or equal to b which is true
```

# Small Project

```python
import time
```

```python
print("device:",*device)
st = time.time()
matrix_a = tt.randint(size=(5000,7000),high = 540,low = -50)
print("matrix a : ", matrix_a)
matrix_b = tt.randint(size=(7000,5000),high=600,low = -999)
print("matrix b:",matrix_b)

```

```python
#checking time taken in cpu of matrix multipulcation of 1100x1000 and 1000x1100
st = time.time()
output = tt.matmul(matrix_a,matrix_b)

et = time.time()
print(output)

print("time taken:",et-st,'seconds')
```
