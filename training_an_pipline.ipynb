```python
import torch #importing module 
import torch.nn as nn #importing another module called nn from torch library
```

```python
#load dataset
with open("input.txt","r",encoding="latin-1") as f:  #loading the input.txt file in read mode
  text = f.read() #keeping that file in a variable called text
text #prining the text variable to see the content of the input.txt file
```

```python
#converting characters into numbers
chars = sorted(set(text)) #converting the text into a set of unique characters and sorting them
print("chars:",chars) #printing the unique characters in the text
stoi = {ch: i for i , ch in enumerate(chars)} #converting back the unique characters into numbers using a dictionary comprehension
print("stoi:",stoi) #printing the dictionary that maps characters to numbers


itos = {i: ch for i , ch in enumerate(chars)} #converting back the numbers into unique characters using a dictionary comprehension
print("itos:",itos) #printing the dictionary that maps numbers to characters
```

```python
data = torch.tensor([stoi[ch] for ch in text],dtype=torch.long) #converting the text into a tensor of numbers
data # printing the tensor of numbers
```

```python
x = data[:-1] # taking all the elements of the tensor except the last one and storing it in a variable called x
print("x:",x)#printing the tensor of numbers
y = data[1:] # taking all the elements of the tensor except the first one and storing it in a variable called y
print("y:",y) # printing the tensor of numbers
```

```python
model = nn.Sequential(  # creating a sequential model,sequential means that the model will be a sequence of layers
    nn.Embedding(len(chars),64), #embedding layer that will convert the input characters into a 64-dimensional vector,embeding means that we will convert the input characters into a vector of numbers
    nn.Linear(64,len(chars)) #linear layer that will convert the 64-dimensional vector into a vector of length equal to the number of unique characters in the text
)
```

```python
criterion = nn.CrossEntropyLoss() # creating a loss function that will calculate the cross-entropy loss between the predicted output and the actual output
optimizer = torch.optim.Adam(model.parameters(),lr = 0.01)# creating an optimizer that will update the parameters of the model using the Adam optimization algorithm with a learning rate of 0.001 

# best tip
#make the learning rate higher than 0.001, because the model is small and the dataset is small, so the model will converge faster with a higher learning rate.
```

```python
training_times = 4000 #increase it for best results, this is the number of times the model will be trained on the input data but it takes more time to train the model for more epochs, so we will keep it at 1000 for now
for epoch in range(training_times): #training the model for 0.1 million epochs
  output = model(x) #getting the output of the model by passing the input x through the model
  loss = criterion(output,y) #calculating the loss between the predicted output and the actual output using the loss function
  optimizer.zero_grad() #setting the gradients of all the parameters in the model to zero before calculating the gradients for the current epoch
  loss.backward()#calculating the gradients of the loss with respect to the parameters in the model using backpropagation
  optimizer.step()#updating the parameters in the model using the gradients calculated in the previous step and the optimization algorithm defined earlier
  if epoch % 100 == 0: #printing the epoch number and the loss value every 100 epochs
    print("epoch:",epoch,"loss:",loss.item()) #printing the epoch number and the loss value every 100 epochs
```

```python
print("output:",x) # printing the output of the model
```

```python
predicted = torch.argmax(output,dim=1) #taking the index of the maximum value in the output tensor along the specified dimension (dim=1) and storing it in a variable called predicted
generated_text = "".join( #joining the characters corresponding to the predicted indices into a single string
    itos[i.item()] for i in predicted #iterating over the predicted tensor and converting each index back to its corresponding character using the itos dictionary
)
print("generated text:",generated_text) #printing the generated text after training the model
```

```python
#calculating the accuracy of the model by comparing the predicted output with the actual output
correct = (predicted == y).sum().item() #calculating the number of correct
print("correct:",correct) #printing the number of correct predictions
total = y.size(0) #calculating the total number of predictions
accuracy = correct / total #calculating the accuracy of the model
print("accuracy:",round(accuracy,4)*100,"%") #printing the accuracy of the model
```
